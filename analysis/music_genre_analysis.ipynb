```python

import pandas as pd
import matplotlib.pyplot as plt
df=pd.read_excel("music dataset - Copy.xlsx")
print("MUSIC GENRE ANALYSIS")
print("Total participants:",len(df))
print("Total columns:",df.shape[1])
print("FAVOURITE GENRE ANALYSIS")
genres=df["Fav genre"].unique()
genre_names=[]
genre_counts=[]
for genre in genres:
    count=0
    for value in df["Fav genre"]:
        if value==genre:
            count=count + 1
    genre_names.append(genre)
    genre_counts.append(count)
    print(genre,":",count)
plt.figure(figsize=(10,5))
plt.bar(genre_names,genre_counts)
plt.title("Favourite Music Genres")
plt.xlabel("Music Genre")
plt.ylabel("Number of Participants")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
print("STREAMING PLATFORM ANALYSIS")
platforms=df["Primary streaming service"].unique()
platform_names=[]
platform_counts=[]
for platform in platforms:
    count=0
    for value in df["Primary streaming service"]:
        if value==platform:
            count=count + 1
    platform_names.append(platform)
    platform_counts.append(count)
    print(platform,":",count)
plt.figure(figsize=(10,5))
plt.bar(platform_names,platform_counts)
plt.title("Most Used Streaming Platforms")
plt.xlabel("Streaming Platform")
plt.ylabel("Number of Participants")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
print("LISTENING HABITS ANALYSIS")
hours = df["Hours per day"].unique()
print("\nListening Hours:")
for hour in sorted(hours):
    count=0
    for value in df["Hours per day"]:
        if value==hour:
            count=count + 1
    print(hour,"hours:",count)
plt.figure(figsize=(8,5))
plt.hist(df["Hours per day"],bins=10,edgecolor="black")
plt.title("Daily Music Listening Hours")
plt.xlabel("Hours per Day")
plt.ylabel("Number of Participants")
plt.tight_layout()
plt.show()
print("\nListening While Working:")
work_options=df["While working"].unique()
for option in work_options:
    count=0
    for value in df["While working"]:
        if value==option:
            count=count+1
    print(option,":",count)
plt.figure(figsize=(6, 5))
work_counts=[]
for option in work_options:
    count=0
    for value in df["While working"]:
        if value==option:
            count=count + 1
    work_counts.append(count)
plt.bar(work_options, work_counts)
plt.title("Listening to Music While Working")
plt.xlabel("Response")
plt.ylabel("Number of Participants")
plt.tight_layout()
plt.show()
```

```python

age_groups=[]
for age in df["Age"]:
    if age>=12 and age<=17:
        age_groups.append("Non-Adults")
    elif age>=18 and age<=40:
        age_groups.append("Young")
    elif age>=41 and age<=54:
        age_groups.append("Mid-Age")
    elif age>=55 and age<=60:
        age_groups.append("Senior")
    elif age>=61 and age<=89:
        age_groups.append("Elderly")
    else:
        age_groups.append("Outside defined groups")
df["Age Group"]=age_groups
print("Age group analysis")
groups=df["Age Group"].unique()
for group in groups:
    count=0
    for value in df["Age Group"]:
        if value==group:
            count=count + 1
    print(group,":",count)
```

```python

print("FAVOURITE GENRE BY AGE GROUP")
for group in groups:
    print("\nAge Group:",group)
    genres=df["Fav genre"].unique()
    highest_count=0
    most_popular=""
    for genre in genres:
        count=0
        for i in range(len(df)):
            if df["Age Group"][i]==group:
                if df["Fav genre"][i]==genre:
                    count=count + 1
        print(genre,":",count)
        if count>highest_count:
            highest_count=count
            most_popular=genre
    print("Most popular genre:",most_popular)
```

```python

age_group_names=[]
popular_genres=[]
popular_counts=[]
for group in groups:
    highest_count=0
    most_popular=""
    for genre in df["Fav genre"].unique():
        count=0
        for i in range(len(df)):
            if df["Age Group"][i]==group:
                if df["Fav genre"][i]==genre:
                    count=count + 1
        if count>highest_count:
            highest_count=count
            most_popular=genre
    if group!="Outside defined groups":
        age_group_names.append(group)
        popular_genres.append(most_popular)
        popular_counts.append(highest_count)
plt.figure(figsize=(9,5))
plt.bar(age_group_names,popular_counts)
plt.title("Most Popular Genre by Age Group")
plt.xlabel("Age Group")
plt.ylabel("Number of Participants")
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()
```
